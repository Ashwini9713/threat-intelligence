```python
import hashlib
import os

# Known malicious file hashes (in practice, pulled from a threat intel feed)
MALWARE_HASHES = {
    "e99a18c428cb38d5f260853678922e03abd8334": "Example.Malware.A",
    # Add more known SHA-256 hashes here
}

def calculate_sha256(file_path):
    sha256_hash = hashlib.sha256()
    try:
        with open(file_path, "rb") as f:
            for chunk in iter(lambda: f.read(4096), b""):
                sha256_hash.update(chunk)
        return sha256_hash.hexdigest()
    except Exception as e:
        print(f"Error reading {file_path}: {e}")
        return None

def scan_directory(directory):
    print(f"Scanning directory: {directory}\n")
    for root, _, files in os.walk(directory):
        for file in files:
            path = os.path.join(root, file)
            file_hash = calculate_sha256(path)
            if file_hash:
                if file_hash in MALWARE_HASHES:
                    print(f"[ALERT] {path}\n  -> Matches known malware: {MALWARE_HASHES[file_hash]}\n  -> Hash: {file_hash}")
                else:
                    print(f"[OK] {path} -> {file_hash}")

if __name__ == "__main__":
    target_dir = input("Enter directory to scan: ")
    scan_directory(target_dir)
```
