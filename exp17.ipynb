## Simple Port Scanner

This notebook demonstrates a basic port scanner in Python. It uses the `socket` library to attempt connections to specified ports on a target host. For improved performance, it leverages `threading` to scan multiple ports concurrently.


```python
import socket
import threading
from queue import Queue
import time

print_lock = threading.Lock()
```

### 1. Port Scanning Function

This function attempts to connect to a given port on the target host. If successful, it prints that the port is open.

```python
def portscan(port):
    s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    s.settimeout(1) # Set a timeout for connection attempts
    try:
        con = s.connect((target, port))
        with print_lock:
            print(f'Port {port}: Open')
        con.close()
    except:
        pass # Port is closed or unreachable, ignore.
```

### 2. Worker Thread Function

This function is executed by each worker thread. It continuously retrieves a port from the queue and calls the `portscan` function for that port.

```python
def worker():
    while True:
        worker = q.get()
        portscan(worker)
        q.task_done()
```

```python
# --- Configuration ---
target = 'localhost' # Replace with the target IP address or hostname
start_port = 1
end_port = 1024 # Common ports to scan
num_threads = 100 # Number of concurrent threads to use

# --- Initialization ---
q = Queue()

print(f"Scanning {target} from port {start_port} to {end_port} with {num_threads} threads...")

# Start timer
t0 = time.time()

# Create and start worker threads
for x in range(num_threads):
    t = threading.Thread(target=worker)
    t.daemon = True # Allows the main program to exit even if threads are still running
    t.start()

# Populate the queue with ports to scan
for port in range(start_port, end_port + 1):
    q.put(port)

# Wait for all tasks in the queue to be completed
q.join()

# End timer
t1 = time.time()

print(f"Scanning completed in {t1-t0:.2f} seconds.")
print("Note: Only 'Open' ports are displayed above.")
```
